```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import plotly.express as px
import numpy as np
import pandas as pd

def confusion_matrix_algo(y_test,y_pred,label_encoder,title):

    # Calculer la matrice de confusion
    conf_matrix = confusion_matrix(y_test, y_pred)

    # Créer un DataFrame à partir de la matrice de confusion
    conf_matrix_df = pd.DataFrame(conf_matrix, index=label_encoder.classes_, columns=label_encoder.classes_)

    # Visualiser la matrice de confusion avec Plotly
    fig = px.imshow(conf_matrix_df, 
                    text_auto=True, 
                    color_continuous_scale='Blues', 
                    labels=dict(x='Classe prédite', y='Classe réelle', color='Nombre de prédictions'),
                    title=title)

    # Ajuster l'affichage des axes
    fig.update_xaxes(side="top")
    fig.show()
```

```python
confusion_matrix_algo(y_test,y_pred,label_encoder,"Matrice de confusion de CART")
```

```python
# Importer les bibliothèques nécessaires
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score
import matplotlib.pyplot as plt
import time
import psutil

# Extraire les caractéristiques et la cible
y = df_attack["label"]
X = df_attack[["n_pkt_dst", "n_pkt_src", "proto", "size", "modbus_response", "flags", "label_n", "ip_s", "mac_s"]]

# Encoder les colonnes en valeurs numériques
mac_encoder = LabelEncoder()
X["mac_s"] = mac_encoder.fit_transform(X["mac_s"])

ip_encoder = LabelEncoder()
X["ip_s"] = ip_encoder.fit_transform(X["ip_s"])

proto_encoder = LabelEncoder()
X["proto"] = proto_encoder.fit_transform(X["proto"])

# Encoder les étiquettes
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Diviser les données en ensembles d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Mesurer le temps d'entraînement
start_fit_time = time.time()
knn_model = KNeighborsClassifier(n_neighbors=5)  # Vous pouvez ajuster le nombre de voisins
knn_model.fit(X_train, y_train)
fit_time = time.time() - start_fit_time

# Mesurer l'utilisation de la RAM après l'entraînement
ram_usage_fit = psutil.Process().memory_info().rss / (1024 * 1024)  # RAM en Mo

# Mesurer le temps de prédiction
start_predict_time = time.time()
y_pred = knn_model.predict(X_test)
predict_time = time.time() - start_predict_time

# Évaluer le modèle
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=label_encoder.classes_)

# Afficher les résultats
print(f"\nAccuracy: {accuracy:.2f}")
print("Classification Report:")
print(report)

# Afficher les temps d'entraînement et de prédiction, ainsi que l'utilisation de la RAM
print(f"\nTime to fit the model: {fit_time:.4f} seconds")
print(f"RAM usage during fitting: {ram_usage_fit:.2f} MB")
print(f"Time to predict: {predict_time:.4f} seconds")
```
