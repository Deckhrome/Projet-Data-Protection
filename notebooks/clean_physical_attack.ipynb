```python
# Load data 
# ONLY FOR phy_att_4.csv the delmiter is ',' and utf-8 encoding
df_1 = pd.read_csv(f'{DATA_DIR}/Physical dataset/phy_att_1.csv', delimiter='\t', header=0, encoding='utf-16')
df_2 = pd.read_csv(f'{DATA_DIR}/Physical dataset/phy_att_2.csv', delimiter='\t', header=0, encoding='utf-16')
df_3 = pd.read_csv(f'{DATA_DIR}/Physical dataset/phy_att_3.csv', delimiter='\t', header=0, encoding='utf-16')
df_4 = pd.read_csv(f'{DATA_DIR}/Physical dataset/phy_att_4.csv', delimiter=',', header=0, encoding='utf-8')
```

```python
# rename for df_2 column name Lable_n to Label_n
df_2.rename(columns={'Lable_n': 'Label_n'}, inplace=True)

# For df_2 change nomal value for Label to normal
df_2['Label'] = df_2['Label'].replace('nomal', 'normal')

all_df = [df_1, df_2, df_3, df_4]
```

# Almost zero NaN values in the physical attacks

```python
def get_columns_info(df,column_name):
    print('Column name: ', column_name)
    print('Number of unique values in attack: ', df[column_name].nunique())
    print('Number of missing values in attack: ', df[column_name].isnull().sum())
    print('Number of values in attack: ', df[column_name].count())
    print('Values in attack: ', df[column_name].unique())
    print('Values count: ', df[column_name].value_counts(normalize=True))
```

```python
print("Process all dateframe...")
for df in all_df:
    # Drop rows with missing values
    df.dropna(inplace=True)
    # Drop duplicate rows
    df.drop_duplicates(inplace=True)
    # Change Time type to datetime[ns]
    df['Time'] = pd.to_datetime(df['Time'])
    # Encoding to category type
    df['Label'] = df['Label'].astype('category')
    # Encoding to boolean type
    df['Label_n'] = df['Label_n'].astype('bool')
print("Done!")
```
