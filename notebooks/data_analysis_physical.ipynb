```python
# Distribution des valeurs pour les colonnes Pump
def plot_pump(df):
    pump_columns = [col for col in df.columns if 'Pump' in col]

    # Calculer la proportion de True/False
    pump_stats = df[pump_columns].mean().reset_index()
    pump_stats.columns = ['Pump', 'Proportion_True']

    fig = px.bar(
        pump_stats,
        x='Pump',
        y='Proportion_True',
        title="Proportion de valeurs True pour les colonnes Pump",
        labels={'Proportion_True': 'Proportion True', 'Pump': 'Colonnes Pump'},
        text_auto=True,
    )
    fig.update_layout(height=500)
    fig.show()
```

```python
plot_pump(df_3)
```

Les répartition False et true sont sensiblement les même pour les 4 dataframes, avec 0 True pour la Pump_3

```python
def plot_flow(df):
    flow_columns = [col for col in df.columns if 'Flow_sensor' in col]
    for col in flow_columns:
        fig = px.histogram(df, x=col, nbins=50, title=f"Distribution des valeurs pour {col}")
        fig.update_layout(bargap=0.1)
        fig.show()
```

```python
plot_flow(df_2)
```

```python
def plot_valve(df):
    valv_columns = [col for col in df.columns if 'Valv' in col]

    # Calculer la proportion de True pour chaque colonne
    valv_stats = df[valv_columns].mean().reset_index()
    valv_stats.columns = ['Valv', 'Proportion_True']

    fig = px.bar(
        valv_stats,
        x='Valv',
        y='Proportion_True',
        title="Proportion de valeurs True pour les colonnes Valv",
        labels={'Proportion_True': 'Proportion True', 'Valv': 'Colonnes Valv'},
        text_auto=True,
    )
    fig.update_layout(height=500)
    fig.show()
```

```python
plot_valve(df_1)
plot_valve(df_2)
plot_valve(df_3)
plot_valve(df_4)
```

Le taux de vrai pour les valve est similaire entre chaque df pour chaque valve sauf pour le df_2

```python
def label_distributions(df):
    # Distribution des labels (Label)
    fig = px.bar(
        df['Label'].value_counts().reset_index(),
        x='count',
        y='Label',
        title="Distribution des labels (Label)",
        labels={'index': 'Classe', 'Label': 'Nombre d’échantillons'},
        text_auto=True,
    )
    fig.update_layout(height=500)
    fig.show()

    # Distribution des classes (Label_n)
    fig = px.bar(
        df['Label_n'].value_counts().reset_index(),
        x='count',
        y='Label_n',
        title="Distribution des classes (Label_n)",
        labels={'index': 'Classe', 'Label_n': 'Nombre d’échantillons'},
        text_auto=True,
    )
    fig.update_layout(height=500)
    fig.show()
```

```python
label_distributions(df_4)
```

```python
# Exemple pour Pump_1
def transition_pump(df,pump):
    df['Pump_1_Transitions'] = df[pump].astype(int).diff().abs()
    transitions = df['Pump_1_Transitions'].sum()

    fig = px.bar(
        x=['Pump_1 Transitions'],
        y=[transitions],
        title="Nombre de transitions (True <-> False) pour Pump_1",
        labels={'x': 'État', 'y': 'Nombre de transitions'},
        text_auto=True,
    )
    fig.show()
```

```python
transition_pump(df_1,'Pump_6')
```
